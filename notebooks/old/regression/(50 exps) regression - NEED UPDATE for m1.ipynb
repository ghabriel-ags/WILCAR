#### <p style='text-align: justify;'>O Método para a obtenção dos pesos iniciais propõe que: (i) seja realizada a linearização, via série de Taylor, do modelo não-linear a ser utilizado para a modelagem do conjunto de dados (o modelo consiste em uma rede neural com p entradas e um neurônio na camada de saída), onde o ponto de equilíbrio para a linearização é a média das suas respectivas variáveis (x,y); (ii) seja realizada a Regressão Linear Múltipla, via Método dos Mínimos Quadrados, do conjunto de dados a ser modelado; (iii) através da comparação matemática entre (i) e (ii), obtenha-se os pesos que conectam os neurônios de entrada ao primeiro neurônio oculto da rede.

Os pesos iniciais do neurônio oculto a ser adicionado são obtidos a partir do método de Bengio e Glorot (2010), conhecido como método de Xavier. ## Definindo função para cálculo de MSE
MSE é a função de perda, utilizado para a otimização do ARP.

```python
def mse(actual, pred): 
    actual, pred = np.array(actual), np.array(pred)
    return np.square(np.subtract(actual,pred)).mean() 
```

### Aplicação de Regressão Linear no conjunto de dados

```python
inputs_all_cte = sm.add_constant(inputs_all, prepend=True) #Adiciona a coluna das constantes
res = sm.OLS(targets_all,inputs_all_cte).fit() #Cria e ajusta o modelo
print (res.summary()) #Imprime os resultados
```

### Resolução do sistema de equações proposto para obter os parâmetros iniciais da rede
<p style='text-align: justify;'>A resolução é feita utilizando o módulo scipy.optimize.fsolve. O módulo retorna as raízes das equações (não lineares) definidas por f(x)= 0, dada uma estimativa inicial.</p>

```python
#Definindo constantes:
x_M = []
for i in range(n_inputs):
    x_M.append(np.mean(inputs_all[:,i]))
k = []
for i in range(n_inputs):
    k.append(res.params[i+1])
b = res.params[0]
```

```python
#Definindo chutes iniciais para beta e wi como valores aleatórios entre 0 e 1:
x0 = np.random.rand(n_inputs+1,1)
```

```python
#Definindo função sigmoide e sua derivada:
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def dsigmoid(x):
    z = sigmoid(x)
    return z * (1 - z)
```

```python
#Inicializando variáveis wi e beta:
w = [0 for i in range(n_inputs)]
beta = 0
```

```python
#Definindo sistema de equações a ser resolvido:
def f(x):
    w,beta = x[:-1],x[-1]
    f = [sigmoid(np.dot(w,x_M)+beta)-(sigmoid(np.dot(w,x_M)+beta)-(sigmoid(np.dot(w,x_M)+beta))**2)*np.dot(w,x_M)-b]
    for i in range(n_inputs):
        f.append((sigmoid(np.dot(w,x_M)+beta)-(sigmoid(np.dot(w,x_M)+beta))**2)*w[i]-k[i])
    return f
```

```python
#Resolvendo o sistema e imprimindo resultados (pesos e bias iniciais):
result = fsolve(f, x0)
print(result)
```

### Criando Método 1:

```python
class Method1:
    def __init__(self, x, y, n):
        
        self.Yh = np.zeros((1,y.shape[1])) #vetor de saídas preditas. Inicialmente um vetor de zeros.
        self.n = n #número de neurônios ocultos.
        self.dims = [x.shape[0], n, 1] #número de neurônios em cada camada (entrada, intermediária e saída).
        self.param = {} #um dicionário que conterá os parâmetros W e b entre as camadas da rede (entrada-intermediária e intermediária-saída); ou seja: W1, b1, W2 e b2.
        self.ch = {} #parâmetro intermediário de cálculo.
        self.grad = {} #parâmetro intermediário de cálculo.
        self.loss = [] #erro de treinamento (MSE).
        self.lr = 0.1 #taxa de aprendizagem.
        self.sam = y.shape[1] #número de observações do conjunto de treinamento.

    def nInit(self): #inicializa os pesos e os bias. Nesse caso, os pesos e os bias do primeiro neurônio oculto corresponde aos pesos encontrados pelo processo acima.
        
        self.param['W1'] = np.random.uniform(-math.sqrt(6)/math.sqrt(self.dims[0]+1),math.sqrt(6)/math.sqrt(self.dims[0]+1),(self.dims[1], self.dims[0]))
        self.param['W1'][0] = result[:-1]
        self.param['b1'] = np.random.uniform(-math.sqrt(6)/math.sqrt(self.dims[0]+1),math.sqrt(6)/math.sqrt(self.dims[0]+1),(self.dims[1], 1))
        self.param['b1'][0] = result[-1]     
        self.param['W2'] = np.random.uniform(-math.sqrt(6)/math.sqrt(self.dims[1]+1),math.sqrt(6)/math.sqrt(self.dims[1]+1),(self.dims[2], self.dims[1]))
        self.param['b2'] = np.random.uniform(-math.sqrt(6)/math.sqrt(self.dims[1]+1),math.sqrt(6)/math.sqrt(self.dims[1]+1),(self.dims[2], 1))
        return

#Definindo função para cálculo de erro quadrático:
    def nloss(self):
        loss = mse(y,self.Yh)
        return loss

#Agora, vamos definir, dentro da classe, uma função que executará o cálculo em cada unidade de cada camada da rede.
#Ela se chamará forward porque ela encaminhará a entrada da rede para a camada oculta que, por sua vez, a encaminhará pela o neurônio de saída.
#Trata-se do sentido direto de propagação de informação através da rede.
#Para tanto, multiplica-se os pesos da primeira camada pelos dados de entrada e adiciona-se a primeira matriz de bias, b1, para produzir Z1.
#Em seguida, aplica-se a função sigmoide em Z1 para produzir A1.
#Após isso, multiplica-se a matriz de pesos da segunda camada pela sua entrada, A1 (a saída da primeira camada, que é a entrada da segunda camada)
#e adicionamos a segunda matriz de viés, b2, para produzir Z2. Em seguida, aplicamos a função sigmoide em Z2 para produzir A2, que é, de fato Yh, 
#a saída final da rede.
    def forward(self,x):
        Z1 = self.param['W1'].dot(x) + self.param['b1']
        A1 = sigmoid(Z1)
        self.ch['Z1'],self.ch['A1'] = Z1,A1
        Z2 = self.param['W2'].dot(A1) + self.param['b2']
        A2 = sigmoid(Z2)
        self.ch['Z2'],self.ch['A2'] = Z2,A2
        self.Yh = A2
#Acabamos de processar nossos dados de entrada pela rede e produzimos Yh, uma saída.
#O próximo passo é descobrir o quão bom foi o nosso resultado.
#Para isso, podemos comparar a saída que produzimos com a saída que deveríamos ter obtido: Yh e Y. Para isso, aplicaremos a função de erro:
        loss = self.nloss()
        return self.Yh, loss

    def backward(self):
        dLoss_Yh = - (y-self.Yh)   
        
        dLoss_Z2 = dLoss_Yh * dsigmoid(self.ch['Z2'])    
        dLoss_A1 = np.dot(self.param["W2"].T,dLoss_Z2)
        dLoss_W2 = 1/self.ch['A1'].shape[1] * np.dot(dLoss_Z2,self.ch['A1'].T)
        dLoss_b2 = 1/self.ch['A1'].shape[1] * np.dot(dLoss_Z2, np.ones([dLoss_Z2.shape[1],1])) 
                            
        dLoss_Z1 = dLoss_A1 * dsigmoid(self.ch['Z1'])        
        dLoss_A0 = np.dot(self.param["W1"].T,dLoss_Z1)
        dLoss_W1 = 1/x.shape[1] * np.dot(dLoss_Z1,x.T)
        dLoss_b1 = 1/x.shape[1] * np.dot(dLoss_Z1, np.ones([dLoss_Z1.shape[1],1]))  
        
        self.param["W1"] = self.param["W1"] - self.lr * dLoss_W1
        self.param["b1"] = self.param["b1"] - self.lr * dLoss_b1
        self.param["W2"] = self.param["W2"] - self.lr * dLoss_W2
        self.param["b2"] = self.param["b2"] - self.lr * dLoss_b2

    def train(self, X, Y, iter = 1501):
        for i in range(0, iter):
            Yh,loss = self.forward(x)
            self.backward()
        self.loss.append(loss)   
        return
    
    def test(self,x,y):
        pred, loss = self.forward(x)
        return loss
```

```python
#Inicializando contagem do tempo:
ini=time.time()

#Abrindo listas para R2 MAX (treino e teste):
r2_train_max = []
r2_test_max = []

#Inicializando loop de experimentos:
for l in range(1,51):
    np.random.seed(l)
    train_inputs, test_inputs, train_targets, test_targets = train_test_split(inputs_all, targets_all, test_size=0.20, random_state=l)
    print("Experimento",l,)
    
    #Abrindo listas para MSE e R2 (treino e teste):
    mse_train = []
    mse_test = []
    r2_train = []
    r2_test = []

    #Inicializando loop para neurônios ocultos:
    for n in range(1, 251):
        
        #Sinalização em tela de início de rede com n neurônios ocultos:
        print("Iniciando rede com",n,"neurônios ocultos")

        #Definindo entrada e saída de treino:
        x = train_inputs.T
        y = np.ndarray.tolist(train_targets)
        y = [y]
        y = np.asarray(y)

        #Criando modelo com n neurônios ocultos:
        Modelo = Method2(x,y,n)

        #Inicializando pesos:
        Modelo.nInit()

        #Treinando modelo:
        Modelo.train(x,y)

        #Registrando e exibindo MSE de treino:
        mse_train.append(Modelo.loss)
        #print("MSE de treino:",Modelo.loss)

        #Registrando e exibindo R2 de treino:
        Yh = [ item for elem in Modelo.Yh for item in elem]
        y = np.ndarray.tolist(train_targets)
        r2_train.append(r2_score(y, Yh))
        #print("R2 de treino:",r2_score(y, Yh))

        #Definindo entrada e saída de teste:
        x = test_inputs.T
        y = np.ndarray.tolist(test_targets)
        y = [y]
        y = np.asarray(y)

        #Testando modelo e registrando MSE de teste:
        mse_test.append(Modelo.test(x,y))

        #Registrando e exibindo R2 de teste:
        Yh = [ item for elem in Modelo.Yh for item in elem]
        y = np.ndarray.tolist(test_targets)
        r2_test.append(r2_score(y, Yh))
        #print("R2 de teste:",r2_score(y, Yh))
    r2_train_max.append(r2_train[r2_test.index(max(r2_test))])   
    r2_test_max.append(max(r2_test))
```
